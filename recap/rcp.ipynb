```python
# logging
import logging

# logging basic configuration
logging.basicConfig(level=logging.DEBUG)

# logging messages
logging.debug("Debugging information")
logging.info("Informational message")
logging.warning("Warning: something is not right")
logging.error("Error: something went wrong")
logging.critical("Critical error: system is down")
```

```python
logging.basicConfig(
    filename='app.log',
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s - %(levelname)s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
    )

# logging messages
logging.debug("Debugging information")
logging.info("Informational message")
logging.warning("Warning: something is not right")
logging.error("Error: something went wrong")
logging.critical("Critical error: system is down")
```

## Flask

```python
from flask import Flask


app=Flask(__name__)

@app.route("/")
def welcome():
    return "welcome to this Flask test"

@app.route("/index")
def index():
    return "welcome to index page"

@app.route("/about")
def about():
    return "welcome to about page"

if __name__=="__main__":
    app.run(
        debug=True,
        use_reloader=False
        )
```
