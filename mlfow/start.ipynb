```python
import mlflow
```

before we need to run mlflow server running with this command:
mlflow ui

```python
mlflow.set_tracking_uri(uri="http://127.0.0.1:5000")
# this means every experiment we are running is running in this url.
```

```python
mlflow.set_experiment(experiment_name="check localhost connection")
```

```python
with mlflow.start_run():
    mlflow.log_metric("test",1)
    mlflow.log_metric("farzam",2)
```
